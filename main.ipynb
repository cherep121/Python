Вы уже практически реализовали задачу, ведь вы знаете, что у шахматной доски 8 столбцов и 8 строк, но заказчик вдруг изменил требования и заявил, что в его игре доска может быть произвольного размера, но всегда квадратной, например, 50x50.

Напишите функцию make_field(size), в которую поступает размер доски, и которая возвращает требуемый массив numpy. Тип элементов массива должен быть np.int8. Для решения задачи нельзя использовать циклы.

```python
import numpy as np

def make_field(size):
    rows = np.arange(size).reshape(-1, 1)
    cols = np.arange(size)
    field = (rows + cols) % 2
    field = 1 - field
    return field.astype(np.int8)

size = 8
board = make_field(size)
print(f"Доска {size}x{size} (1 - белая, 0 - черная):")
print(board)

size = 5
board = make_field(size)
print(f"\nДоска {size}x{size}:")
print(board)

print(f"\nПроверка: левая верхняя клетка = {board[0, 0]} (ожидается 1)")
```

Для решения задачи нельзя использовать циклы. Для сортировки используйте функцию np.sort.

```python
import numpy as np

def super_sort(rows, cols):
    A = np.random.randint(1, 101, (rows, cols))
    B = A.copy()
    
    B[:, ::2] = np.sort(B[:, ::2], axis=0)
    
    B[:, 1::2] = np.sort(B[:, 1::2], axis=0)[::-1, :]
    
    return (A, B)

rows, cols = 5, 6
A, B = super_sort(rows, cols)

print(f"Матрица А (оригинал) {rows}x{cols}:")
print(A)
print(f"\nМатрица B (после сортировки):")
print(B)

print("\nПроверка:")
print(f"Четные столбцы (индексы 0,2,4): отсортированы по возрастанию вдоль строк")
print(f"Нечетные столбцы (индексы 1,3,5): отсортированы по убыванию вдоль строк")
```

№5 (2 балла) Создайте две матрицы размера n×n и заполните их случайными значениями. Посчитайте их поэлементное и матричное произведения. Для этого реализуйте функции:

    mult – матричное перемножение без использования numpy

    el_mult – поэлементное перемножение без использования numpy

    np_mult – матричное перемножение с использованием numpy

    np_el_mult – поэлементное перемножение с использованием numpy

При помощи timeit засеките время работы функций для n = 100 000 (для чистоты эксперимента засекайте время на одинаковых матрицах). Во сколько раз numpy-версия функций работает быстрее реализаций без numpy?

```python
import numpy as np
import time

def mult(A, B):
    n = len(A)
    result = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                result[i][j] += A[i][k] * B[k][j]
    return np.array(result)

def el_mult(A, B):
    n = len(A)
    result = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            result[i][j] = A[i][j] * B[i][j]
    return np.array(result)

def np_mult(A, B):
    return np.dot(A, B)

def np_el_mult(A, B):
    return A * B

print("\nПроверка корректности на матрице 3x3:")
A_test = np.array([[1, 2], [3, 4]])
B_test = np.array([[5, 6], [7, 8]])

print("Матрица A:")
print(A_test)
print("Матрица B:")
print(B_test)

result_manual = mult(A_test, B_test)
result_numpy = np_mult(A_test, B_test)
print(f"\nРезультат умножения (вручную):\n{result_manual}")
print(f"Результат умножения (numpy):\n{result_numpy}")
print(f"Результаты совпадают: {np.allclose(result_manual, result_numpy)}")

print("\n" + "=" * 50)
print("Замер времени для матриц 200x200:")
n = 200
A = np.random.rand(n, n)
B = np.random.rand(n, n)

start = time.time()
result_manual_mat = mult(A, B)
time_manual_mat = time.time() - start

start = time.time()
result_numpy_mat = np_mult(A, B)
time_numpy_mat = time.time() - start

print(f"Матричное умножение вручную: {time_manual_mat:.4f} сек")
print(f"Матричное умножение с numpy: {time_numpy_mat:.4f} сек")
print(f"numpy быстрее в {time_manual_mat/time_numpy_mat:.0f} раз")

start = time.time()
result_manual_el = el_mult(A, B)
time_manual_el = time.time() - start

start = time.time()
result_numpy_el = np_el_mult(A, B)
time_numpy_el = time.time() - start

print(f"\nПоэлементное умножение вручную: {time_manual_el:.4f} сек")
print(f"Поэлементное умножение с numpy: {time_numpy_el:.4f} сек")
print(f"numpy быстрее в {time_manual_el/time_numpy_el:.0f} раз")
```

№9 (2 балла) Вычислите результат формулы:

```python
import numpy as np

X = np.array([[-3, 4, 1], [4, 3, 1]])
y = np.array([[10], [12]])
lam = 0.1

print("Матрица X:")
print(X)
print(f"\nВектор y: {y.ravel()}")
print(f"λ = {lam}")

I_3 = np.eye(3)
print(f"\nЕдиничная матрица I₃:")
print(I_3)

X_T = X.T
first_part = X_T @ X + lam * I_3
print(f"\nXᵀ·X + λ·I =")
print(first_part.round(4))

inverse_part = np.linalg.inv(first_part)
print(f"\n(Xᵀ·X + λ·I)⁻¹ =")
print(inverse_part.round(4))

theta = inverse_part @ X_T @ y
print(f"\nθ = (Xᵀ·X + λ·I)⁻¹·Xᵀ·y =")
print(theta)

print(f"\nРезультат: θ₁ = {theta[0,0]:.4f}, θ₂ = {theta[1,0]:.4f}, θ₃ = {theta[2,0]:.4f}")
```

№10 (2 балла) Напишите программу, которая будет решать систему линейных уравнений вида:

```python
import numpy as np

print("Система вида:")
print("a11·x + a12·y = b1")
print("a21·x + a22·y = b2")
print()

print("\nПример 1: Система с единственным решением")
print("2x + 3y = 8")
print("3x + 4y = 11")

A1 = np.array([[2, 3], [3, 4]])
B1 = np.array([8, 11])
det1 = np.linalg.det(A1)
print(f"Определитель: {det1}")

if abs(det1) > 1e-10:
    solution1 = np.linalg.solve(A1, B1)
    print(f"Решение: x = {solution1[0]}, y = {solution1[1]}")
else:
    print("Система не имеет решения или имеет бесконечно много решений")

print()
print("Пример 2: Система без решения (определитель = 0)")
print("2x + 4y = 8")
print("1x + 2y = 5")

A2 = np.array([[2, 4], [1, 2]])
B2 = np.array([8, 5])
det2 = np.linalg.det(A2)
print(f"Определитель: {det2}")

if abs(det2) > 1e-10:
    solution2 = np.linalg.solve(A2, B2)
    print(f"Решение: x = {solution2[0]}, y = {solution2[1]}")
else:
    print("Определитель равен 0 → система не имеет решения или имеет бесконечно много решений")

print()
print("Интерактивный режим (введите свои коэффициенты):")
print()

try:
    a11 = float(input("Введите a11: "))
    a12 = float(input("Введите a12: "))
    b1 = float(input("Введите b1: "))
    a21 = float(input("Введите a21: "))
    a22 = float(input("Введите a22: "))
    b2 = float(input("Введите b2: "))
    
    A = np.array([[a11, a12], [a21, a22]])
    B = np.array([b1, b2])
    det = np.linalg.det(A)
    
    if abs(det) < 1e-10:
        print("\nОпределитель = 0 → система не имеет решения или имеет бесконечно много решений")
    else:
        solution = np.linalg.solve(A, B)
        print(f"\nРешение: x = {solution[0]:.4f}, y = {solution[1]:.4f}")
except ValueError:
    print("Ошибка ввода")
except KeyboardInterrupt:
    print("\nПрограмма завершена")
```

№11 (2 балла) Используя numpy, без питоновских циклов сконвертируйте цветное изображение в чёрно-белое по формуле

```python
import numpy as np
from PIL import Image

def bw_convert(input_path='input_image.jpg', output_path='output_bw.jpg'):
    try:
        image = Image.open(input_path)
        print(f"Исходное изображение: {image.size}, {image.mode}")
        
        img_array = np.array(image)
        
        if len(img_array.shape) != 3 or img_array.shape[2] != 3:
            print("Ошибка: изображение не в формате RGB")
            return False
        
        R = img_array[:, :, 0].astype(float)
        G = img_array[:, :, 1].astype(float)
        B = img_array[:, :, 2].astype(float)
        
        gray = 0.2989 * R + 0.5870 * G + 0.1140 * B
        
        gray = np.round(gray).astype(np.uint8)
        
        result_image = Image.fromarray(gray, mode='L')
        result_image.save(output_path)
        
        print(f"Черно-белое изображение сохранено как {output_path}")
        print(f"Размер: {result_image.size}, режим: {result_image.mode}")
        return True
        
    except FileNotFoundError:
        print(f"Ошибка: файл {input_path} не найден")
        print("Создаю тестовое изображение для демонстрации...")
        
        test_img = Image.new('RGB', (400, 300), color='red')
        test_img.save(input_path)
        print(f"Создан тестовый файл {input_path}")
        return bw_convert(input_path, output_path)

bw_convert()
```

Для чтения данных из этого файла в объект numpy.ndarray воспользуйтесь методом genfromtxt():

```python
import numpy as np

try:
    table = np.genfromtxt('ABBREV.csv', delimiter=';', dtype=None, 
                          names=True, encoding='utf8', filling_values=0)
    
    print(f"Загружено {len(table)} продуктов")
    print(f"Доступные столбцы: {list(table.dtype.names)[:10]}...") 
    
    all_columns = list(table.dtype.names)
    
    calories_col = None
    sugar_col = None
    protein_col = None
    vitc_col = None
    name_col = None
    
    for col in all_columns:
        if 'Kcal' in col:
            calories_col = col
        if 'Sugar' in col:
            sugar_col = col
        if 'Protein' in col:
            protein_col = col
        if 'Vit_C' in col:
            vitc_col = col
        if 'Shrt_Desc' in col or 'Desc' in col:
            name_col = col
    
    print(f"\nНайденные столбцы:")
    print(f"  Калории: {calories_col}")
    print(f"  Сахар: {sugar_col}")
    print(f"  Белок: {protein_col}")
    print(f"  Витамин C: {vitc_col}")
    print(f"  Название: {name_col}")
    
    print("\n" + "-" * 50)
    max_calories = np.max(table[calories_col])
    max_calorie_foods = table[table[calories_col] == max_calories]
    max_calorie_names = max_calorie_foods[name_col]
    max_calorie_food = np.sort(max_calorie_names)[-1] if len(max_calorie_names) > 0 else max_calorie_names[0]
    print(f"★ Самый калорийный продукт:")
    print(f"   Название: {max_calorie_food}")
    print(f"   Калорийность: {max_calories} ккал")
    
    print("\n" + "-" * 50)
    if sugar_col:
        sugar_data = table[sugar_col]
        sugar_data_clean = np.where(np.isnan(sugar_data), np.inf, sugar_data)
        valid_sugar = sugar_data_clean < 1e10
        if np.any(valid_sugar):
            min_sugar = np.min(sugar_data_clean[valid_sugar])
            min_sugar_foods = table[(sugar_data_clean == min_sugar) & valid_sugar]
            min_sugar_names = min_sugar_foods[name_col]
            min_sugar_food = np.sort(min_sugar_names)[0] if len(min_sugar_names) > 0 else min_sugar_names[0]
            print(f"★ Самый полезный (минимальный сахар):")
            print(f"   Название: {min_sugar_food}")
            print(f"   Содержание сахара: {min_sugar} г")
    
    print("\n" + "-" * 50)
    if protein_col:
        max_protein = np.max(table[protein_col])
        max_protein_foods = table[table[protein_col] == max_protein]
        max_protein_names = max_protein_foods[name_col]
        max_protein_food = np.sort(max_protein_names)[0] if len(max_protein_names) > 0 else max_protein_names[0]
        print(f"★ Самый богатый белком продукт:")
        print(f"   Название: {max_protein_food}")
        print(f"   Содержание белка: {max_protein} г")
    
    print("\n" + "-" * 50)
    if vitc_col:
        vit_c_data = table[vitc_col]
        vit_c_data_clean = np.where(np.isnan(vit_c_data), -np.inf, vit_c_data)
        valid_vit_c = vit_c_data_clean > -1e10
        if np.any(valid_vit_c):
            max_vit_c = np.max(vit_c_data_clean[valid_vit_c])
            vit_c_foods = table[(vit_c_data_clean == max_vit_c) & valid_vit_c]
            if len(vit_c_foods) > 0:
                vit_c_food = vit_c_foods[name_col][0]
                print(f"★ Самый богатый витамином C продукт:")
                print(f"   Название: {vit_c_food}")
                print(f"   Содержание витамина C: {max_vit_c} мг")
    
except FileNotFoundError:
    print("Ошибка: файл ABBREV.csv не найден")
    print("Убедитесь, что файл находится в той же папке, что и программа")
except Exception as e:
    print(f"Ошибка при обработке файла: {e}")
    print("\nПопробуем вывести названия всех столбцов для отладки:")
    try:
        with open('ABBREV.csv', 'r', encoding='utf8') as f:
            first_line = f.readline().strip()
            columns = first_line.split(';')
            print("Столбцы в файле:")
            for i, col in enumerate(columns):
                print(f"  {i}: '{col}'")
    except:
        pass
```

№14 (1 балл) Напишите функцию reverse(), которая побайтно читает бинарные данные из файла input.dat и создает файл output.dat, куда сохраняет прочитанные данные в обратном порядке.

Постарайтесь решить эту задачу без использования циклов.

```python
def reverse():

    try:
        with open('input.dat', 'rb') as infile:
            data = infile.read()
        
        reversed_data = data[::-1]
        
        with open('output.dat', 'wb') as outfile:
            outfile.write(reversed_data)
        
        return True
    except FileNotFoundError:
        print("Файл input.dat не найден")
        return False

test_data = b'Hello, World! This is a test file.'
with open('input.dat', 'wb') as f:
    f.write(test_data)

print(f"Исходные данные: {test_data}")
print(f"Длина: {len(test_data)} байт")

if reverse():
    with open('output.dat', 'rb') as f:
        result = f.read()
    print(f"\nРезультат: {result}")
    print(f"Обратный порядок: {result == test_data[::-1]}")
```

№15 (1 балл) В текстовом файле input.txt через пробельные символы (пробел, табуляция, перевод строки и т.д.) записаны целые числа — данные, полученные с некоторого устройства. Количество их неизвестно.

```python
import re

test_content = "1 -1 2 -2 0 4 9 100 8"
with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(test_content)

print(f"Содержимое input.txt: {test_content}")

with open('input.txt', 'r', encoding='utf-8') as f:
    content = f.read()

numbers = list(map(int, re.findall(r'-?\d+', content)))

print(f"\nНайденные числа: {numbers}")
print(f"Всего чисел: {len(numbers)}")

positive = [n for n in numbers if n > 0]
negative = [n for n in numbers if n < 0]
zero = [n for n in numbers if n == 0]

print(f"Положительных: {len(positive)}")
print(f"Отрицательных: {len(negative)}")
print(f"Нулей: {len(zero)}")

with open('output.txt', 'w', encoding='utf-8') as f:
    f.write(f"{len(numbers)}\n")
    
    result_parts = []
    if positive:
        result_parts.extend(['1', str(len(positive))])
    if negative:
        result_parts.extend(['-1', str(len(negative))])
    if zero:
        result_parts.extend(['0', str(len(zero))])
    
    output = ' '.join(result_parts)
    f.write(output)

print(f"\nРезультат сохранен в output.txt:")
print(f"Строка 1: {len(numbers)}")
print(f"Строка 2: {output}")

print("\n" + "-" * 50)
print("Проверка с примером 1:")
example1 = "1 -1 2 -2 0 4 9 100 8"
print(f"Ввод: {example1}")
print("Ожидаемый вывод:")
print("9")
print("1 5 -1 2 0 1")
```

№17 (1 балл) В csv-файле хранится информация о продуктах: название, старая и новая цены. Цены указаны в копейках.

Выведите на экран в порядке упоминания в файле те товары, цена на которые уменьшилась.

Если товаров, попадающих под условие задачи, нет, то на экран выводить ничего не надо.

Для быстрого решения задачи воспользуйтесь объектом DictReader.

```python
import csv
import os

test_file = 'prices.csv'
test_data = [
    ['name', 'old_price', 'new_price'],
    ['Товар А', '10000', '9000'],
    ['Товар Б', '5000', '5500'], 
    ['Товар В', '3000', '3000'], 
    ['Товар Г', '20000', '15000'],
    ['Товар Д', '8000', '8000']
]

if not os.path.exists(test_file):
    with open(test_file, 'w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        writer.writerows(test_data)
    print(f"Создан тестовый файл {test_file}")

print("\nТовары с уменьшившейся ценой:")
print("-" * 40)

found = False
with open(test_file, 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        old_price = int(row['old_price'])
        new_price = int(row['new_price'])
        if new_price < old_price:
            print(f"  {row['name']}: {old_price} коп → {new_price} коп")
            found = True

if not found:
    print("  Нет товаров с уменьшившейся ценой")
```
